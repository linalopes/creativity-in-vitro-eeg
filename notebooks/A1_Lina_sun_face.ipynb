```python
from gtts import gTTS

# Generate audio "Imagine the sun on your face"
tts1 = gTTS(text="Imagine the sun on your face", lang='en')
tts1.save("sun.mp3")

# Generate audio "Relax"
tts2 = gTTS(text="Relax", lang='en')
tts2.save("rest.mp3")

print("✅ Audio files saved as sun.mp3 and rest.mp3")
```

```python
from pydub import AudioSegment

# Load the two generated files
sun = AudioSegment.from_file("sun.mp3", format="mp3")
rest = AudioSegment.from_file("rest.mp3", format="mp3")
pause = AudioSegment.silent(duration=4000)  # 4s pause

# Build sequence with 30 cycles
sequence = AudioSegment.silent(duration=1000)  # initial 1s margin
for _ in range(30):
    sequence += sun + pause + rest + pause

# Export final audio
sequence.export("training_sequence.wav", format="wav")
print("✅ Final audio generated as training_sequence.wav")
```

```python
# Load the files
sun = AudioSegment.from_file("sun.mp3", format="mp3")
rest = AudioSegment.from_file("rest.mp3", format="mp3")

# Duration in milliseconds
sun_duration_ms = len(sun)
rest_duration_ms = len(rest)

# Show in ms and also in seconds with 3 decimal places
print(f"🟡 SUN   = {sun_duration_ms} ms = {sun_duration_ms/1000:.3f} s")
print(f"🟢 REST  = {rest_duration_ms} ms = {rest_duration_ms/1000:.3f} s")
```

```python
# Load the final audio
audio = AudioSegment.from_file("training_sequence.wav", format="wav")

# Total duration
duration_ms = len(audio)
duration_sec = duration_ms / 1000

print(f"🎧 training_sequence.wav duration:")
print(f"→ {duration_ms} ms")
print(f"→ {duration_sec:.3f} seconds")
```
